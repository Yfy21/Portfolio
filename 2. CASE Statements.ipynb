Cloud computing is proving to be one of the fundamental pillars of the 4th Industrial Revolution: data centers have become essential assets in the current economical landscape and, in fact, strategic even in foreign affairs. This is because they allow organizations to access technological services such as computing power, storage, and databases maintained and provided by specialized companies on demand, rather than requiring the purchase and maintenance of the physical infrastructure needed to run data centers and servers themselves — and this can SUBSTANTIALLY reduce costs.

One way to efficiently use cloud services is to leverage the scalable computing resources offered in order to reduce the amount of transfered data in any given operation. With regard to SQL, this means that, whenever possible, we should use the available cloud computing resources to **preprocess the data** before downloading it to be used in pipeline stages that do not rely on the cloud. One way of doing this is by aggregating data — which, as we have seen in the previous entry, takes advantage of the GROUP BY clause. However, our data is not always in a format that allows for immediate grouping. Enter the CASE statement.

With the CASE statement, as you can see in the code below, we can relabel data according to certain conditions that can be verified through conditional logic, that is, through if-then statements. This allow us to significantly **reduce the cardinality** of a column by establishing ranges of values and then group by such ranges. In the example below, I have grouped 41 distinct values of `hours_studied` per student per week into 4 different ranges, then aggregated the data using those ranges, significantly reducing the number of rows in the output and avoiding statistics at an overly granular level, which would not be helpful.

## Analysis' Context

The code you'll see is a product of a [Datacamp](https://www.datacamp.com/) challenge in which my goal was to analyze factors that impact in student performance. There are a number of ways to analyze this — if I was using Python, for example, I could use the package [Scikit-Learn](https://scikit-learn.org) to calculate feature importance by running regressions —, but the code you'll see follows one specific instruction: to group students in four pre-determined ranges and to see how this alone affects their performance.

![Student Performance](<images/2nd Entry/student_performance table.png>)

```python
SELECT CASE
		 WHEN hours_studied <= 5 THEN '1-5 hours'
		 WHEN hours_studied <= 10 THEN '6-10 hours'
		 WHEN hours_studied <= 15 THEN '11-15 hours'
		 WHEN hours_studied >= 16 THEN '16+ hours'
	   END AS hours_studied_range,
	   AVG(exam_score) AS avg_exam_score
  FROM student_performance
 GROUP BY hours_studied_range
 ORDER BY avg_exam_score DESC;
```
