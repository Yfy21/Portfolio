## About this Entry

In the previous entry to this portfolio, I showed how Pandas can be leveraged to consolidate data spread across multiple files and in different formats. However, data will not always be locally available to us, which means that we must also be able to fetch it from online resources. One way of doing that is with the [Requests](https://pypi.org/project/requests/) package: a library that allows clients to send HTTP requests to servers and receive their responses.

### HTTP
The Internet is a global system of interconnected computer networks that uses the [Internet Protocol Suite](https://en.wikipedia.org/wiki/Internet_protocol_suite) to communicate between networks and devices. Through it, we can access the World Wide Web (or simply "the Web"), the collection of web pages that make information available in an user-friendly way.

The rules (or protocols) for content-sharing via the Web are set by the Hypertext Transfer Protocol (HTTP), an [application layer protocol](https://en.wikipedia.org/wiki/Application_layer) that regulates how clients and servers must interact in order to transfer information to one another. So, whenever we use a web browser to access content online, what is happening behind the scenes is that we are acting as clients sending requests to servers and waiting for their response. Once we get that response — usually an [HTML](https://en.wikipedia.org/wiki/HTML) file —, the browser's engine converts its content into an interactive visual representation of the page on our device.

### The Requests Package
We don't always need a visual representation of a website's data. In fact, if we are looking to automate our data pipeline, we don't want the data retrieval to be dependent on a human user capable of interacting with a visual representation of the HTML file's content. What we want is for our code to automatically access the relevant pieces of data and process them for our needs.
The Requests package is useful precisely for that! With it, we can send requests to servers via URLs (Uniform Resource Locators) and collect any useful data in their responses; its ability to concatenate parameters in [query strings](https://en.wikipedia.org/wiki/Query_string) is also extremely useful for automating customizable data extraction.

## Analysis' Context
In this entry, I will be using Requests to access and process data on the book *Moby Dick*, available at [Project Gutenberg](https://www.gutenberg.org/). As with other entries in this portfolio, such task was presented as a challenge at https://www.datacamp.com/. The goal is to analyze which words (excluding stop words) appear most in the book.

```python
# Import and download packages
import requests
from bs4 import BeautifulSoup
import nltk
from collections import Counter
nltk.download('stopwords')
```

```python
# Request and prettify data
r = requests.get('https://www.gutenberg.org/cache/epub/2701/pg2701.txt')
r.encoding = 'utf-8'
html = r.text
soup = BeautifulSoup(html, 'html.parser')
moby_text = soup.get_text(separator=' ', strip=True)
```

```python
# Tokenize text
tokenizer = nltk.tokenize.RegexpTokenizer(r'\w+')
tokens = tokenizer.tokenize(moby_text)
words = [*map(str.lower, tokens)]
stop_words = nltk.corpus.stopwords.words('english')
words_no_stop = [word for word in words if word not in stop_words]
```

```python
# Count most frequent words
count = Counter(words_no_stop)
top_ten = count.most_common(10)
top_ten
```
